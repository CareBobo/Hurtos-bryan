# Análisis Exploratorio del Dataset de Hurtos

Este notebook reproduce el diagnóstico preliminar y profundiza en algunos análisis descriptivos que pueden ser incluidos en el artículo científico. No se modifican los datos originales.


```python
import pandas as pd
import plotly.express as px

# Ruta al archivo CSV (mantener sin cambios)
csv_path = '../data/raw/Reporte_Hurto_por_Modalidades_Policía_Nacional_20261106_v2.csv'
df = pd.read_csv(csv_path, encoding='utf-8')
df.head()
```

```python
# Descripción básica del dataset
df.describe(include='all')
```

```python
# Distribución de hurtos por departamento
dept_counts = df['DEPARTAMENTO'].value_counts().reset_index()
dept_counts.columns = ['DEPARTAMENTO', 'conteo']
fig = px.bar(dept_counts, x='DEPARTAMENTO', y='conteo', title='Hurtos por Departamento')
fig.show()
```

```python
# Distribución temporal (si la columna FECHA HECHO es de tipo fecha)
df['FECHA HECHO'] = pd.to_datetime(df['FECHA HECHO'], dayfirst=True, errors='coerce')
time_counts = df['FECHA HECHO'].dt.to_period('M').value_counts().sort_index().reset_index()
time_counts.columns = ['Mes', 'conteo']
fig2 = px.line(time_counts, x='Mes', y='conteo', title='Hurtos por Mes')
fig2.show()
```

## Observaciones preliminares
- Se observa que el dataset contiene registros de los departamentos **ATLÁNTICO**, **META**, **NORTE DE SANTANDER**, **ANTIOQUIA** y **CALDAS**.
- Las modalidades de hurto están representadas por los valores `ARTÍCULO 239. HURTO AUTOMOTORES` y `ARTÍCULO 239. HURTO MOTOCICLETAS`.
- El rango temporal abarca los meses de enero a marzo de 2014.
- No se encontraron valores nulos ni registros duplicados.
